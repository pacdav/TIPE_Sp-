```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('Mesure/V2_1-036.csv')
plt.plot(df['Temps'], np.abs(df['EA0']))
T= df['Temps']
E= np.abs(df['EA0'])
```

```python
def smooth(y, box_pts):
    box = np.ones(box_pts) / box_pts
    y_smooth = np.convolve(y, box, mode="same")
    return y_smooth
#plt.plot(T, E)
#plt.plot(T, smooth(E, 3))
plt.plot(T, smooth(E, 100))
plt.grid()
```
