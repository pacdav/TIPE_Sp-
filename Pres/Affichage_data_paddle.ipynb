```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Données
annees = ['2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025', '2026', '2028', '2030']
valeurs = [0.95, 1.10, 1.45, 1.70, 1.55, 1.58, 1.65, 1.76, 1.88, 2.15, 2.44]

# Configuration du style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 8)) # Légèrement plus large pour l'équilibre

# Création des barres
# J'ai gardé ta couleur verte #8BA888 qui rappelle bien l'univers "nature/outdoor" du SUP
bars = plt.bar(annees, valeurs, color='#8BA888', edgecolor='#6B8469', linewidth=1.2, alpha=0.9)

# --- PERSONNALISATION DU TITRE (PLUS GROS ET PLUS HAUT) ---
plt.title('ÉVOLUTION DU MARCHÉ MONDIAL DU SUP (2018 - 2030)', 
          fontsize=22, 
          pad=35, 
          fontweight='black', 
          color='#1D3557', 
          loc='center')

# Ajout des étiquettes de données
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.04, f'{yval} Md$', 
             ha='center', va='bottom', fontsize=11, fontweight='bold', color='#1D3557')

# Personnalisation des axes
plt.xlabel('Année', fontsize=14, labelpad=15, fontweight='semibold')
plt.ylabel('Taille du marché (Milliards USD)', fontsize=14, labelpad=15, fontweight='semibold')

# Amélioration des graduations
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.ylim(0, 2.8)

# Suppression des bordures inutiles pour un look moderne
sns.despine(left=True, bottom=True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Données
annees = ['2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025', '2026', '2028', '2030']
valeurs = [0.95, 1.10, 1.45, 1.70, 1.55, 1.58, 1.65, 1.76, 1.88, 2.15, 2.44]

# Création de la figure
plt.figure(figsize=(10, 6))

# Création des barres
# On utilise tes codes couleurs hexadécimaux pour la cohérence
bars = plt.bar(annees, valeurs, color='#556d96', edgecolor='#556d96', linewidth=1.2)

# Personnalisation esthétique
plt.title('Marché Mondial du SUP (Milliards USD)', fontsize=14, fontweight='bold', color='#333333', pad=20)
plt.ylabel('Valeur en Md $', fontsize=11)
plt.xlabel('Années', fontsize=11)

# Ajout d'une grille horizontale discrète
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Ajout des valeurs au-dessus des barres
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.03,
             f'{height}', ha='center', va='bottom', fontsize=10, fontweight='bold')

# Suppression des bordures superflues (top et right) pour un look moderne
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Ajustement pour éviter que les étiquettes ne soient coupées
plt.tight_layout()

# Affichage
plt.show()
```
